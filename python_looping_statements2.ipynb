```python
'''1. Print Numbers from 1 to n
Question: Write a program to print numbers from 1 to n. 
Explanation: Use a loop starting from 1 to n and print each number. - Input: n = 5 - Output: 1 2 3 4 5
'''
n = int(input())
for i in range(1,n+1):
    print(i, end=' ')
```

```python
'''11. Prime Number Check
Question: Check if a number is prime. 
Explanation: A number is prime if it has exactly 2 factors. - Input: n = 7 - Output: Prime
'''
n = int(input())

if n <= 1:
  print("Not Prime")
else:
  is_prime = True
  for i in range(2, n):
    if n % i == 0:
      is_prime = False
      break

  if is_prime:
    print("Prime")
  else:
    print("Not Prime")
```

```python
'''19. Armstrong Number Check
Question: Check if a number is an Armstrong number. 
Explanation: Sum of cube of digits equals the number. - Input: 153 - Output: Armstrong number
'''
n = int(input())
d = n
sum = 0
while d>0:
    sum+=(d%10)**3
    d//=10
if n == sum:
    print("Armstrong number")
else:
    print("Not an Armstrong number")
```

```python
'''22. Count Vowels in String
Question: Count number of vowels in a string. 
Explanation: Loop and check for a, e, i, o, u. - Input: “apple” - Output: 2
'''
s = input()
count = 0
vowels = ['a','e', 'i','o','u', 'A','E', 'I','O','U']
for i in (s):
    if i in vowels:
        count+=1
print(count)
```

```python
'''24. Count Vowels and Consonants
Question: Count vowels and consonants in input string. 
Explanation: Maintain two counters. - Input: “apple” - Output: Vowels = 2, Consonants = 3
'''
s = input()
vow = 0
con = 0
vowels = ['a','e', 'i','o','u', 'A','E', 'I','O','U']
l1 = list(range(65,91))
l2 = list(range(97,123))
l = l1+l2
for i in s:
    if ord(i) in (l):
        if i in vowels:
            vow+=1
        else:
            con+=1
print(f"Vowels count in {s}: {vow}")
print(f"Constants count in {s}: {con}")
```

```python
'''25. Perfect Number Check
Question: Check if a number is perfect. 
Explanation: Sum of proper divisors equals the number. - Input: 28 - Output: Perfect number
'''
n = int(input())
divisors_sum = 1
if n <= 1:
    print('give the number greather than 1')
else:
    for i in range(2,int(n**0.5) + 1):
        if n%i == 0:
            divisors_sum += i
            if (n//i) != i:
                divisors_sum += (n//i)
    if divisors_sum == n:
        print("Perfect number")
    else:
        print("Not a perfect number")
```

```python
'''26. Neon Number Check
Question: Check if a number is a neon number. 
Explanation: Square the number, sum digits, match original. - Input: 9 - Output: Neon number
'''
n = int(input())
square = n**2
sum = 0
while square>0:
    sum+=square%10
    square//=10
if sum==n:
    print("Neon number")
else:
    print("Not a neon number")
```

```python
'''27. Strong Number Check
Question: Check if a number is a strong number. 
Explanation: Sum of factorial of digits equals the number. - Input: 145 - Output: Strong number
'''
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

n = int(input())
temp = n
total_sum = 0

if n == 0:
    total_sum = factorial(0)
else:
    while temp > 0:
        digit = temp % 10
        total_sum += factorial(digit)
        temp //= 10

if total_sum == n:
    print("Strong number")
else:
    print("Not a strong number")
```

```python
'''29. Fibonacci Series
Question: Print the Fibonacci series up to n terms. 
Explanation: Start with 0, 1 and continue with sum of last two. - Input: n = 5 - Output: 0 1 1 2 3
'''
n = int(input())
a, b = 0, 1
if n <= 0:
    print("Please enter a positive integer")
elif n == 1:
    print(a)
else:
    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
```

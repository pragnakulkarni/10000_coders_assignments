```python
'''1. Square, Rectangle, and Triangle Patterns (1–15)
Solid Square Pattern
Problem: Print a solid square of stars of size n.
Input: n = 4
Output:
* * * *
* * * *
* * * *
* * * *
'''
n = int(input())
for i in range(n):
    for j in range(n):
        print('*',end = ' ')
    print()
```

```python
'''2. Solid Rectangle Pattern
Problem: Print a solid rectangle of m rows and n columns.
Input: m = 3, n = 5
Output:
* * * * *
* * * * *
* * * * *
'''
m, n = int(input()), int(input())
for i in range(m):
    for j in range(n):
        print('*', end = ' ')
    print()
```

```python
'''3. Right-Angled Triangle (Left-Aligned)
Problem: Print a left-aligned right-angled triangle.
Input: n = 5
Output:
*
* *
* * *
* * * *
* * * * *
'''
n = int(input())
i=1
while i<=n:
    for j in range(i):
        print('*', end = ' ')
    print()
    i+=1
```

```python
'''5. Inverted Triangle (Left-Aligned)
Input: n = 5
Output:
* * * * *
* * * *
* * *
* *
*
'''
n = int(input())
i=n
while i >=1:
    for j in range(i):
        print('*', end=' ')
    print()
    i-=1
```

```python
'''6. Inverted Triangle (Right-Aligned)
Input: n = 5
Output:
* * * * *
  * * * *
    * * *
      * *
        *
'''
n = int(input())
for i in range(n):
    print('  '*i + '* '*(n-i))
```

```python
'''7. Centered Pyramid Pattern
Input: n = 4
Output:
      *
    * * *
  * * * * *
* * * * * * *
'''
n = int(input())
for i in range(1,n+1):
    for j in range(i,n):
        print('  ', end = '')
    m=i+(i-1)
    print('* ' * m)
```

```python
'''8. Diamond Pattern
Input: n = 3
Output:
    *
  * * *
* * * * *
  * * *
    *
'''
n = 3
for i in range(1,n+1):
    for j in range(i,n):
        print('  ', end = '')
    m=i+(i-1)
    print('* ' * m)
temp = n-1
for i in range(1,temp+1):
    print('  '*i, end = '')
    m = (temp - i + 1) + (temp - i)
    print('* ' * m)
```

```python
'''9. Butterfly Pattern
Input: n = 4
Output:
*       *
* *   * *
* * * * *  
* *   * *
*       *
'''
#written the code for standard butterfly, given pattern is slightly different from the standard one
def print_butterfly(n):
    # Top Half
    for i in range(1, n + 1):
        stars = "* " * i
        spaces = "  " * (2 * (n - i))
        print(stars + spaces + stars)
    
    # Bottom Half
    for i in range(n - 1, 0, -1):
        stars = "* " * i
        spaces = "  " * (2 * (n - i))
        print(stars + spaces + stars)

# Example usage
n = 3
print_butterfly(n)
```

```python
'''16. Hollow Square Pattern
Problem: Print a hollow square of stars of size n.
Input: n = 4
Output:
* * * *
*     *
*     *
* * * *
'''
n = int(input())
for i in range(n):
    for j in range(n):
        if i == 0 or i == n-1 or j == 0 or j == n-1:
            print('*', end = ' ')
        else:
            print(' ', end = ' ')
    print()
```

```python
'''22. Hollow Pyramid Pattern
Input: n = 4
Output:
      *
    *   *
  *       *
* * * * * * *
'''
n = int(input())
for i in range(n):
    if i == 0:
        print('  ' * (n-1) + '*')
    elif i == n-1:
        print('* ' * (2 * i) + '*')
    else:
        print('  ' * (n-i-1) + '* ' + '  ' * (i+i-1) + '*')
```

```python
'''23. Hollow Diamond Pattern
Input: n = 3
Output:
    *
  *   *
*       *
  *   *
    *
'''
n = int(input())
k = n+n-1
m1 = n-2
m2 = 1
inc = 1
dec = k-4
for i in range(k):
    if i == 0 or i == k-1:
        print('  ' * (n-1) + '*')
    elif i == k//2:
        print('* ' + '  ' * (k-2) + '*')
    elif i < k//2:
        print('  ' * m1 + '* ' + '  ' * inc + '*')
        m1 -= 1
        inc += 2
    else:
        print('  ' * m2 + '* ' + '  ' * dec + '*')
        m2+=1
        dec -= 2
```

```python
'''24. Hollow Butterfly Pattern
Input: n = 4
Output:
*       *
* *   * *
*   *   *
*       *
*   *   *
* *   * *
*       *
'''
n = 4
for i in range(1, n+1):
    mid_spaces = '  ' * (2*(n-i))
    if i == 1:
        print('* ' + mid_spaces + '+ ')
    else:
        hollow_spaces = '  ' * (i-2)
        print('* ' + hollow_spaces + '* ' + mid_spaces + '* ' + hollow_spaces + '* ')
for j in range(1, n+1):
    mid_spaces = '  ' * (2*(j-1))
    if j<n:
        hollow_spaces = '  ' * ((n-1)-j)
        print('* ' + hollow_spaces + '* ' + mid_spaces + '* ' + hollow_spaces + '* ')
    else:
        print('* ' + mid_spaces + '* ')
```

```python
'''25. Hollow Hourglass Pattern
Input: n = 5
Output:
* * * * *
*       *
  *   *
    *
  *   *
*       *
* * * * *
'''
n = int(input("Enter n: "))

# Upper half
for i in range(n, 0, -1):
    spaces = n - i

    if i == n:
        print("* " * i)
    elif i == 1:
        print(" " * spaces + "*")
    else:
        print(" " * spaces + "*" + " " * (2 * i - 3) + "*")

# Lower half
for i in range(2, n + 1):
    spaces = n - i

    if i == n:
        print("* " * i)
    else:
        print(" " * spaces + "*" + " " * (2 * i - 3) + "*")
```

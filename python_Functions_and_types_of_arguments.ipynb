```python
'''1. Print All Prime Numbers from m to n
Problem: Given a range from m to n, print all prime numbers in that range.
Input: m = 10, n = 30
Output: 11 13 17 19 23 29
Explanation: A prime number has only two factors: 1 and itself.'''
def prime(m,n):
    if m > n:
        print("Invalid range: m should be less than or equal to n")
        return []
    p = []
    for i in range(m,n+1):
        if i<2:
            continue
        factors = [1,i]
        for j in range(2,int(i**0.5) + 1):
            if i%j==0:
                factors.append(j)
                break
        if len(factors) == 2:
            p.append(i)
    return(p)
    
m = int(input())
n = int(input())
print(*prime(m,n))
```

```python
'''2. Count of All Prime Numbers from m to n
Problem: Count how many prime numbers are there between m and n.
Input: m = 1, n = 10
Output: 4
Explanation: Prime numbers are: 2, 3, 5, 7'''
def prime(m,n):
    if m > n:
        print("Invalid range: m should be less than or equal to n")
        return []
    p = []
    for i in range(m,n+1):
        if i<2:
            continue
        factors = [1,i]
        for j in range(2,int(i**0.5) + 1):
            if i%j==0:
                factors.append(j)
                break
        if len(factors) == 2:
            p.append(i)
    return(len(p))
    
m = int(input())
n = int(input())
print(f"count of prime number between {m} and {n} are: ", prime(m,n))
```

```python
'''3. Print All Armstrong Numbers in a Range
Problem: Print all Armstrong numbers between m and n.
Input: m = 1, n = 500
Output: 1 153 370 371 407
Explanation: Armstrong number = sum of each digit raised to the power of number of digits.'''
def armstrong(a):
    num_str = str(a)
    l = len(num_str)
    s = sum(int(i)**l for i in num_str)
    if a == s:
        return 1
    else:
        return 0

def all_armstrong(m,n):
    if m > n:
        print("Invalid range: m should be less than or equal to n")
        return []
    l = []
    for i in range(m,n+1):
        if armstrong(i) == 1:
            l.append(i)
    return l
    
m = int(input())
n = int(input())
print(*all_armstrong(m,n))
```

```python
'''4. First Prime Number from m to n
Problem: Find the first prime number in the given range.
Input: m = 10, n = 25
Output: 11'''
def first_prime(m,n):
    if m > n:
        print("Invalid range: m should be less than or equal to n")
        return None
    for i in range(m,n+1):
        if i<2:
            continue
        factors = [1,i]
        for j in range(2,int(i**0.5) + 1):
            if i%j == 0:
                factors.append(j)
                break
        if len(factors) == 2:
            return(i)
    return None
m = int(input())
n = int(input())
result = first_prime(m,n)
if result is not None:
    print(f"first prime number between {m} and {n} is: {result}")
else:
    print(f"No prime numbers found between {m} and {n}.")
```

```python
'''5. Last Prime Number from m to n
Problem: Find the last prime number in the given range.
Input: m = 10, n = 25
Output: 23'''
def last_prime(m,n):
    if m > n:
        print("Invalid range: m should be less than or equal to n")
        return None
    for i in range(n,m-1,-1):
        if i<2:
            continue
        is_prime = True
        for j in range(2,int(i**0.5) + 1):
            if i%j == 0:
                is_prime = False
                break
        if is_prime:
            return(i)
    return None
m = int(input())
n = int(input())
result = last_prime(m,n)
if result is not None:
    print(f"last prime number between {m} and {n} is: {result}")
else:
    print(f"No prime numbers found between {m} and {n}.")
```

```python
'''6. First Vowel in a Name
Problem: Given a string, find the first vowel in the string.
Input: name = "Krishna"
'''
def first_vowel(name):
    vowels = ['a','e', 'i','o','u', 'A','E', 'I','O','U']
    for i in name:
        if i in vowels:
            return i
    return None
name = input()
result = first_vowel(name)
if result is not None:
    print(f"First vowel in the name is: {result}")
else:
    print(f"{name} does not have any vowels")
```

```python
'''7. Last Vowel in a Name
Problem: Given a string, find the last vowel in the string.
Input: name = "Ramakrishna"
Output: a
Explanation: Last vowel is ‘a’'''
def last_vowel(name):
    vowels = ['a','e', 'i','o','u', 'A','E', 'I','O','U']
    for i in range(-1,-len(name)-1,-1):
        if name[i] in vowels:
            return name[i]
    return None
name = input()
result = last_vowel(name)
if result is not None:
    print(f"Last vowel in the name is: {result}")
else:
    print(f"{name} does not have any vowels")
```

```python
'''8. Print All Even Numbers Using Continue
Problem: Use continue statement to skip odd numbers and print only even numbers between 1 and n.
Input: n = 10
Output: 2 4 6 8 10'''
def print_even_numbers(n: int):
    if n <= 0:
        print("n should be greater than 0")
        return

    for i in range(1, n + 1):
        if i % 2 != 0:
            continue
        print(i, end=" ")
    print()  # Print a clean newline at the end

# Driver Code
n = int(input())
print_even_numbers(n)
```

```python
'''9. Print All Odd Numbers Using Continue
Problem: Use continue statement to skip even numbers and print only odd numbers.
Input: n = 10
Output: 1 3 5 7 9'''
def print_odd_numbers(n: int):
    if n <= 0:
        print("n should be greater than 0")
        return

    for i in range(1, n + 1):
        if i % 2 == 0:
            continue
        print(i, end=" ")
    print()  # Print a clean newline at the end

# Driver Code
n = int(input())
print_odd_numbers(n)
```

```python
'''10. Count of Prime and Composite Numbers from m to n
Problem: Count how many are prime and how many are composite numbers in range m to n.
Input: m = 1, n = 10
Output: Prime: 4, Composite: 4
Explanation: Prime: 2,3,5,7 | Composite: 4,6,8,9'''
def count_prime_composite(m,n):
    if m > n:
        print("Invalid range: m should be less than or equal to n")
        return (0,0)
    prime = 0
    composite = 0
    for i in range(m,n):
        if i<2:
            continue
        factors = 2
        for j in range(2,int(i**0.5) + 1):
            if i%j==0:
                factors += 1
                break
        if factors == 2:
            prime += 1
        else:
            composite += 1
    return(prime, composite)
    
m = int(input())
n = int(input())
prime_count,composite_count= count_prime_composite(m,n)
print(f"Prime: {prime_count} | Composite: {composite_count}")
```

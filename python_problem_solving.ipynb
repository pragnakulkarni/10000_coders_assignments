```python
'''25. Perfect Number Check
Question: Check if a number is perfect. 
Explanation: Sum of proper divisors equals the number. - Input: 28 - Output: Perfect number
'''
n = int(input())
divisors_sum = 1
if n <= 1:
    print('give the number greather than 1')
else:
    for i in range(2,int(n**0.5) + 1):
        if n%i == 0:
            divisors_sum += i
            if (n//i) != i:
                divisors_sum += (n//i)
    if divisors_sum == n:
        print("Perfect number")
    else:
        print("Not a perfect number")
```

```python
'''27. Strong Number Check
Question: Check if a number is a strong number. 
Explanation: Sum of factorial of digits equals the number. - Input: 145 - Output: Strong number
'''
def factorial(n):
    if n <= 1:
        return 1
    return n * factorial(n - 1)

n = int(input())
temp = n
total_sum = 0

if n == 0:
    total_sum = factorial(0)
else:
    while temp > 0:
        digit = temp % 10
        total_sum += factorial(digit)
        temp //= 10

if total_sum == n:
    print("Strong number")
else:
    print("Not a strong number")
```

```python
'''9. Butterfly Pattern
Input: n = 4
Output:
*       *
* *   * *
* * * * *  
* *   * *
*       *
'''
#written the code for standard butterfly, given pattern is slightly different from the standard one
def print_butterfly(n):
    # Top Half
    for i in range(1, n + 1):
        stars = "* " * i
        spaces = "  " * (2 * (n - i))
        print(stars + spaces + stars)
    
    # Bottom Half
    for i in range(n - 1, 0, -1):
        stars = "* " * i
        spaces = "  " * (2 * (n - i))
        print(stars + spaces + stars)

# Example usage
n = 3
print_butterfly(n)
```

```python
'''24. Hollow Butterfly Pattern
Input: n = 4
Output:
*       *
* *   * *
*   *   *
*       *
*   *   *
* *   * *
*       *
'''
n = 4
for i in range(1, n+1):
    mid_spaces = '  ' * (2*(n-i))
    if i == 1:
        print('* ' + mid_spaces + '+ ')
    else:
        hollow_spaces = '  ' * (i-2)
        print('* ' + hollow_spaces + '* ' + mid_spaces + '* ' + hollow_spaces + '* ')
for j in range(1, n+1):
    mid_spaces = '  ' * (2*(j-1))
    if j<n:
        hollow_spaces = '  ' * ((n-1)-j)
        print('* ' + hollow_spaces + '* ' + mid_spaces + '* ' + hollow_spaces + '* ')
    else:
        print('* ' + mid_spaces + '* ')
```

```python
'''22. Find Frequency of All Elements
Problem: Return a dictionary with the frequency of each element. Input: [1, 2, 2, 3, 1] Output: {1: 2, 2: 2, 3: 1}
'''
def frequency(list_):
    ele_freq = {}
    for i in list_:
        ele_freq[i] = list_.count(i)
    return ele_freq
list_ = list(map(int,input().split()))
res = frequency(list_)
print(res)
```

```python
'''23. Flatten a Nested List
Problem: Convert a nested list into a single list. Input: [[1, 2], [3, 4]] Output: [1, 2, 3, 4]
'''
def flatten_list(list_):
    return [i for j in list_ for i in j]
list_ = [[1,2],[3,4]]
res = flatten_list(list_)
print(res)
```

```python
'''24. Split a List into Even and Odd Lists
Problem: Separate even and odd numbers into different lists. 
Input: [1, 2, 3, 4, 5] Output: Even: [2, 4], Odd: [1, 3, 5]
'''
def even_list(list_):
    return [i for i in list_ if i%2 == 0]
def odd_list(list_):
    return [i for i in list_ if i%2 != 0]
list_ = [1,2,3,4,5]
even, odd = even_list(list_), odd_list(list_)
print(f"Even: {even}, Odd: {odd}")
```

```python
'''25. Find Pair of Elements with Given Sum
Problem: Find all pairs in the list whose sum equals a given value. 
Input: [1, 2, 3, 4], sum = 5 Output: [(1, 4), (2, 3)]
'''
list_ = list(map(int, input().split()))
sum = int(input())
output = []
for i in range(len(list_)):
    t = ()
    for j in range(i+1,len(list_)):
        if list_[i] + list_[j] == sum:
            t += (list_[i],list_[j])
            output.append(t)
print(output)
```

```python
'''38. Rotate List Elements Right
Problem: Rotate list by k positions to the right. Input: [1, 2, 3, 4], k = 2 Output: [3, 4, 1, 2]
'''
def rotate_right(lst, k):
    if not lst:
        return lst
    k = k % len(lst)
    return lst[-k:] + lst[:-k]
nums = list(map(int,input().split()))
k = int(input())
print(rotate_right(nums, k))
```

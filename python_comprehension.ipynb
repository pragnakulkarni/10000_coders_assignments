# Comprehension Assignment: Applying Lambda Functions and Functional Programming Concepts
Objective
This assignment is designed to help students practice and reinforce their understanding of lambda functions, map(), filter(), and sorted() in Python, applying these concepts to real-world problems and data manipulation tasks.

Background / Context
The assignment focuses on applying lambda functions and other functional programming concepts to solve a variety of tasks, including data transformations, filtering, and sorting. Students will work with different types of data, such as numbers and names, to demonstrate their comprehension of these concepts

```python
'''Tasks / Requirements
Implement the following lambda functions:
Double a given number.'''
n = int(input())
res = lambda n: n * 2
print(f'Double of a number n:{n} is {res(n)}')
```

```python
'''Find the remainder when one number is divided by another.'''
a,b = int(input()), int(input())
remainder = lambda a,b : a%b
print(f"remainder = {remainder(a,b)}")
```

```python
'''Check whether a number is divisible by 5.'''
n = int(input())
divisibility_check = lambda n : (True if n%5 == 0 else False)
print(f'{n} is divisible by 5')
```

```python
#Find the smaller of two numbers.
a, b = int(input()), int(input())
minimum = lambda a,b : min(a,b)
print(f'minimu of two numbers a:{a} and b:{b} is {minimum(a,b)}')
```

```python
#Convert Celsius into Fahrenheit using the formula: (C × 9/5) + 32.
celsius = int(input())
fahrenheit = lambda n : int((n * (9/5)) + 32)
print(f"{fahrenheit(celsius)}°F")
```

```python
'''Use map() and lambda to:
Add 5 to every number in a given list: numbers = [10, 20, 30, 40, 50].'''
numbers = list(map(int, input().split()))
new_numbers = list(map(lambda i: i+5, numbers))
print(new_numbers)
```

```python
'''Convert all names to uppercase in a given list: 
names = ["aparna", "ravi", "priya", "kiran"].'''
names = list(map(str, input().split()))
upper_names = list(map(lambda i: i.upper(), names))
print(upper_names)
```

```python
'''Find ages 18 and above in a given list: ages = [12, 18, 25, 15, 30, 10, 22].'''
ages = list(map(int, input().split()))
adult_ages = list(filter(lambda i: i>=18, ages))
print(adult_ages)
```

```python
'''Find names having more than 4 characters in a given list: 
names = ["Ram", "Aparna", "Ravi", "Priyanka", "Sai"].'''
names = list(map(str, input().split()))
long_names = list(filter(lambda i: len(i) > 4, names))
print(long_names)
```

```python
'''Use sorted() and lambda to sort students based on marks from highest to lowest given a list of tuples:
students = [("Aparna", 85), ("Ravi", 72), ("Priya", 95), ("Kiran", 65)].'''
students = [("Aparna", 85), ("Ravi", 72), ("Priya", 95), ("Kiran", 65)]
sorted_students = sorted(students, key=lambda student: student[1], reverse=True)
print(sorted_students)
```
